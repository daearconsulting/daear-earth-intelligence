# Module 2 — Regional Coverage (scaffold)
**Status:** planned, not yet built out.

## Planned regions (the "Three Rivers" framing from the SHIELD notes)
- Cache la Poudre watershed, Colorado Front Range — **built in this
  portfolio** (`earth_intelligence_toolkit.regions.POUDRE_CAMERON_PEAK`)
- Black Hills watershed(s), South Dakota
- A Tribal watershed in South Dakota Tribal Nations territory (candidate
  regions overlap with the Oceti Sakowin Integrated Data Cube contract area)

## Approach
Each region needs only a `Region` entry in
`earth_intelligence_toolkit.regions` (see `POUDRE_CAMERON_PEAK` for the
pattern) — every function in `data_access.py` and `indicators.py` already
takes a bbox as a plain argument, so extending coverage is a matter of
defining new regions, not new code.
