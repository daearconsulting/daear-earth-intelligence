# Module 4 — Time-Series View (scaffold)
**Status:** planned, not yet built out.

## Purpose
Track the Module 3 composite resilience index over time (annually,
pre-fire through multi-year post-fire recovery) to show the disturbance
and recovery arc explicitly, rather than a single before/after snapshot.

## Approach
`cube.stack_time_series` (already implemented, demonstrated in
`wildfire-landscape-intelligence/notebooks/03_post_fire_recovery.ipynb`)
applied to the Module 3 composite-index calculation across multiple years
of optical scenes — the pattern already exists, this module would apply it
one level up (to the composite index rather than to raw NDVI).

## Planned output
A single line chart: composite resilience index, Cache la Poudre region,
2019 (pre-fire) through present — the visual that turns "we can compute a
resilience score" into "here's how resilience actually moved through a
real disturbance event."
