# Module 1 Indicator Catalog (scaffold)
**Status:** planned, not yet built out.

## Purpose
A single normalized registry of every indicator produced across the
portfolio, so `climate-resilience-indicators` can combine them without each
repo needing to know about the others' internals.

## Planned contents
| Indicator | Source repo | Toolkit function |
|---|---|---|
| NDVI / fuel proxy | wildfire-landscape-intelligence | `indicators.ndvi`, `indicators.fuel_proxy` |
| dNBR / burn severity | wildfire-landscape-intelligence | `indicators.dnbr`, `indicators.burn_severity_class` |
| Erosion susceptibility | wildfire-landscape-intelligence, soil-watershed-intelligence | `indicators.erosion_susceptibility` |
| Soil vulnerability | soil-watershed-intelligence | `indicators.soil_vulnerability` |
| Watershed resilience index | soil-watershed-intelligence | `indicators.watershed_resilience_index` |
| Tribal-land exposure | tribal-wildfire-intelligence | (planned) |

This notebook would formalize that table into a small registry object
(e.g. a dataclass per indicator with units, valid range, and source
function) rather than leaving it as documentation only.
