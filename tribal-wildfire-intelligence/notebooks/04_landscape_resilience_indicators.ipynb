# Module 3 Landscape Resilience Indicators (scaffold)
**Status:** planned, not yet built out.

## Approach
Applies the same `indicators.watershed_resilience_index` composite built in
`soil-watershed-intelligence` Module 3, computed specifically for
Tribal-land sub-areas, and released through the governance layer (Module 4)
at whatever tier is appropriate for the specific Tribal Nation's data-sharing
agreement with Daear.

## Planned outputs
- Resilience index map, Tribal-land extent
- Time-series resilience trend, if historical scenes are available

## Toolkit functions this will use
`indicators.watershed_resilience_index` (already implemented, demonstrated
in `soil-watershed-intelligence/notebooks/03_fire_soil_water_impacts.ipynb`),
`governance.gated_release` (already implemented, demonstrated in Module 4)
