# Module 2 — Wildfire Exposure Analysis (scaffold)
**Status:** planned, not yet built out.

## Data sources (live)
- Tribal boundary polygons (BIA)
- Infrastructure layers (roads, buildings — e.g. Microsoft Building Footprints, OSM)
- Ecosystem/land-cover layers (NLCD)

## Planned outputs
- Exposure index: fraction of Tribal land area / infrastructure / ecosystem
  types within historical or modeled fire perimeters
- Comparison against the erosion-susceptibility and burn-severity layers
  already built in `wildfire-landscape-intelligence`, reused via
  `earth-intelligence-toolkit`

## Toolkit functions this will use
`data_access.get_burn_severity`, `indicators.erosion_susceptibility`
(already implemented) applied to Tribal boundary geometries once real
boundary polygons are wired in
