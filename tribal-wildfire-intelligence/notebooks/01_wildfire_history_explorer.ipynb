# Module 1 — Tribal Wildfire History Explorer (scaffold)
**Status:** planned, not yet built out.

## Data sources (live)
- MTBS historical fire perimeters (multi-decade)
- ICS-209 incident reports (see [[lise]]'s ICS-209+ database work — a
  natural internal collaboration point rather than building this from
  scratch)
- Historical trend analysis: fire frequency/size by decade near Tribal
  boundaries

## Planned outputs
- Interactive timeline of fires intersecting Tribal lands over the historical record
- Frequency/severity trend chart by decade

## Governance note
Any output referencing precise locations near culturally-sensitive sites
would be released through `governance.aggregate_to_public` (see Module 4)
rather than at full resolution.
